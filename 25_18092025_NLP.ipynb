# Types of Tokenization
1. Tokens with 2 words = Bigrams
2. Tokens with 3 words = Trygrams
3. Tokens with n words = ngrams

```python
from nltk.util import bigrams, trigrams, ngrams
```

```python
string = 'we are learner of prakash senapati from 10:30am batch'
quotes_tokens = nltk.word_tokenize(string)
quotes_tokens
```

```python
len(quotes_tokens)
```

```python
quotes_bigrams = list(nltk.bigrams(quotes_tokens))
quotes_bigrams

# create list of 2 consecutive words
```

```python
quotes_trigram = list(nltk.trigrams(quotes_tokens))
quotes_trigram
```

```python
quotes_ngram = list(nltk.ngrams(quotes_tokens, 4))
quotes_ngram

# it's the collection of more than 3 words, here you'll need to pass a parameter
# of how many words you want in a list.
```

# 🔨 Types of Stemmers in NLTK  

---

## 1. Porter Stemmer  
📌 Oldest & most famous stemmer (from 1980 by Martin Porter).  
- Works with a set of rules to chop suffixes like *ing, ly, ed, s*, etc.  

**Examples:**  
- `caresses → caress`  
- `ponies → poni`  
- `caressed → caress`  

✅ **Pros:**  
- Simple, fast, and widely used in older IR (information retrieval) systems.  

❌ **Cons:**  
- Sometimes over-stems or under-stems, leading to weird roots (`ponies → poni`).  

**Use case:**  
- When you want a lightweight, fast stemmer in search engines or small NLP tasks.  

---

## 2. Lancaster Stemmer (Paice-Husk Stemmer)  
📌 Very aggressive stemmer — chops words heavily.  
- Uses an iterative algorithm with lots of rules.  

**Examples:**  
- `caresses → caress`  
- `ponies → pony`  
- `maximum → maxim`  
- `compute → comput`  
- `computer → comput`  
- `computation → comput`  

✅ **Pros:**  
- Produces very short and consistent stems.  

❌ **Cons:**  
- Too aggressive → may reduce words so much that meaning is lost.  

**Use case:**  
- Rarely used in modern NLP, but useful if you want extreme reduction (e.g., compact indexing in search engines).  

---

## 3. Snowball Stemmer (aka Porter2)  
📌 Developed by the same Martin Porter later as an improvement over Porter.  
- More refined and consistent than Porter.  
- Supports multiple languages (English, French, German, Spanish, etc.).  

**Examples (English):**  
- `caresses → caress`  
- `ponies → poni` (like Porter)  
- `running → run`  

✅ **Pros:**  
- More accurate than Porter.  
- Supports many languages.  
- Balanced between speed and accuracy.  

❌ **Cons:**  
- Slightly larger and slower than Porter, but still very efficient.  

**Use case:**  
- Modern NLP applications where you need reliable stemming in multiple languages.  

---

## ⚡ Comparison Summary  

| Stemmer              | Aggressiveness         | Accuracy | Speed     | Best Used For                        |
|----------------------|------------------------|----------|----------|--------------------------------------|
| **Porter**           | Moderate               | Decent   | Fast     | Simple English tasks, legacy systems |
| **Lancaster**        | High (chops a lot)     | Low      | Very Fast| Compact indexing, experimental use   |
| **Snowball (Porter2)** | Balanced              | High     | Fast     | Modern NLP, multilingual projects    |

---

## 🏆 Which is Most Optimized?  
- **SnowballStemmer (Porter2)** is considered the most optimized in practice:  
  - Better accuracy than Porter  
  - Still lightweight and fast  
  - Multilingual support  

👉 That’s why most modern libraries (like search engines, text mining pipelines) use **SnowballStemmer**.  


# 🌱 Stemming vs. Lemmatization in NLP  

## 🔹 Stemming  
- **Definition:** Reduces a word to its root form by chopping off prefixes/suffixes.  
- **Method:** Rule-based, does not use a dictionary or context.  
- **Result:** Can produce incomplete or non-real words.  

**Examples:**  
- `studies → studi`  
- `studying → study`  
- `cats → cat`  
- `better → better`  

**Pros:**  
- Very fast ⚡  
- Lightweight and simple  

**Cons:**  
- May produce non-words (`studi`).  
- Ignores grammar and context.  

---

## 🔹 Lemmatization  
- **Definition:** Reduces a word to its dictionary form (lemma).  
- **Method:** Uses a dictionary + part-of-speech (POS) tagging.  
- **Result:** Always produces valid words.  

**Examples:**  
- `studies → study`  
- `studying → study`  
- `cats → cat`  
- `better → good`  

**Pros:**  
- More accurate ✅  
- Always gives valid words  
- Context-aware (POS tagging)  

**Cons:**  
- Slower 🐢  
- Requires dictionary + POS tagging  

---

## ⚡ Quick Comparison  

| Feature        | **Stemming**                 | **Lemmatization**              |
|----------------|------------------------------|--------------------------------|
| **Output**     | Root stem (can be non-word) | Real dictionary word (lemma)   |
| **Speed**      | Faster ⚡                    | Slower 🐢                      |
| **Accuracy**   | Lower                        | Higher (context-aware)         |
| **Method**     | Rule-based suffix chopping   | Dictionary + POS tagging       |
| **Example**    | *studies → studi*            | *studies → study*              |

---

## 🏆 When to Use  
- **Stemming** → Use when speed matters and you only need rough grouping (e.g., search engines, indexing).  
- **Lemmatization** → Use when accuracy and meaning matter (e.g., chatbots, ML preprocessing, text analytics).  

---

👉 **In short:**  
- **Stemming = fast but rough ✂️**  
- **Lemmatization = slow but smart 🧠**

# NER (Named entity recognition)
- It recognizes the famous entities like google, US president, etc.


```python
from nltk import ne_chunk

NE_sent = 'The US president stays in the Whitehouse'
```

```python
NE_tokens = word_tokenize(NE_sent)
NE_tokens
```

```python
NE_tags = nltk.pos_tag(NE_tokens)
NE_tags
```

```python
NE_ner = ne_chunk(NE_tags)
print(NE_ner)
```
