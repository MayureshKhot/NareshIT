# Lambda Function
- Also called as anonymous function.
- Also assign the funciton to a variable, e.g.
  - f = lambda a: a*a
- lambda is a keyword

```python
f = lambda a: a*a
result = f(10)
print(result)
```

```python
f = lambda a,b : a+b

result = f(2,3)
result
```
