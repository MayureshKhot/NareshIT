# Gradient Descent

**Gradient decent is an optimization algorithm to reach global minima by reducing the learning rate to find the best fit line.**

- So here's what I understood about gradient descent.
- Let's say we start the with the origin as value for C (intercept)
- Now the error is high (obviously), so we'll elevate the line a few points from the origin, like mentioned in the left image.
- We'll keep doing the same, and when calculated in terms of cost function or error or Sum of squared residuals (right diagram), it'll looke like a ball going from up to down then up again.
- and the bottom ball(point) in the gradient decent is called as global minima.
- And if you notice the step count (distance) from 1st point to second point vs 2nd to 3rd, the later one's distance is less than the prior one and it goes continuously for the next points as well, and global minima points has least difference between the immediate prior and immediate next point.
- And that is called **Rate of Learning** and **Momentum**

## SGD (Stochastic Gradient Descent)

- **Meaning that I understood**
  - In gradient descent we work on each and every datapoint so takes more computation, time, etc.
  - what stochastic gd does: it takes mean of datapoints in the clusters, like explained in the image below.
  - It is also an optimization algorithm to reach global minima by clustering the datapoints and taking the mean of those clusters instead of the whole datapoints.
  - Drawback: it takes randomely, so it's messed a bit
