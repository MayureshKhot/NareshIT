# 📘 Data Engineering & System Design — From Basics to Advanced

---

## 1. **What is Data?**

**Definition:**  
Data is raw, unprocessed information. It can be numbers, text, images, videos, etc., that can be stored and used for analysis or decision-making.

**Types of Data:**
- **Quantitative:** Numbers (e.g., temperature, sales).
- **Qualitative:** Text or categories (e.g., reviews, gender).

**Real-World Use Case:**
- GPS location from your phone = data
- Purchase history on Amazon = data

**Interview Questions:**
- What is the difference between data and information?
- What are the different types of data?

---

## 2. **What is a Database?**

**Definition:**  
A **database** is an organized collection of data, stored and accessed electronically using a Database Management System (DBMS).

**Common Types:**
- **Relational (SQL)**: Structured data in tables.
- **Non-Relational (NoSQL)**: Flexible data storage formats.

**Real-World Use Case:**
- Banks store customer account information in a database.
- Spotify stores playlists and user preferences.

**Interview Questions:**
- What are ACID properties?
- Compare SQL vs NoSQL databases.

---

## 3. **ETL (Extract, Transform, Load)**

**Definition:**
A common data integration process to move and clean data:
- **Extract**: Pull data from various sources.
- **Transform**: Clean or reshape the data.
- **Load**: Store it in a target system (database, data warehouse).

**Real-World Use Case:**
- A company gathers sales data from stores (extract), standardizes the format (transform), and loads it into a reporting dashboard (load).

**Interview Questions:**
- How would you optimize an ETL pipeline?
- What tools have you used for ETL? (e.g., Apache NiFi, Airflow, Talend)

---

## 4. **Structured vs Unstructured Data**

| Type              | Description                         | Examples                        |
|-------------------|-------------------------------------|---------------------------------|
| **Structured**     | Tabular, organized in rows & columns | SQL, Excel, CSV                 |
| **Unstructured**   | No fixed format or structure         | Images, PDFs, social media posts|

**Semi-Structured** (Bonus):  
- JSON, XML – Not fully structured but has markers.

**Interview Questions:**
- How would you process unstructured data?
- What tools are used for handling large volumes of unstructured data?

---

## 5. **Relational Database (RDBMS)**

**Definition:**  
Stores data in related tables using rows and columns. Uses SQL for querying.

**Key Features:**
- Relationships (foreign keys)
- ACID compliance
- Schemas

**Examples:**  
MySQL, PostgreSQL, Oracle, SQL Server

**Real-World Use Case:**
- A school links student records with their grades and attendance using relational tables.

**Interview Questions:**
- What are normalization and denormalization?
- What is the difference between `JOIN`, `UNION`, and `SUBQUERY`?

---

## 6. **Vector Database**

**Definition:**  
A database optimized to store and search high-dimensional **vectors**. Mainly used in AI/ML to enable similarity search.

**Example Use Cases:**
- Chatbots searching for relevant answers
- Image and video similarity search
- Recommendation engines

**Popular Tools:**  
Pinecone, Weaviate, FAISS, Milvus

**Interview Questions:**
- What is a vector index?
- How does a vector database differ from traditional databases?

---

## 7. **Database vs Server vs Cloud**

| Term      | Definition |
|-----------|------------|
| **Database** | A structured system to store and query data. |
| **Server**   | A physical/virtual machine that hosts applications/databases. |
| **Cloud**    | Internet-based services (e.g., AWS, Azure) that offer scalable servers and databases. |

**Important Clarification:**  
❌ A server is **not** a collection of databases.  
✅ A server **hosts** one or more databases or services.

**Interview Questions:**
- What are the benefits of deploying databases in the cloud?
- Explain horizontal vs vertical scaling.

---

## 8. **Server-Side Concepts**

### a. **Application Server**
- Hosts backend code and business logic.
- Processes user requests and interacts with the database.

**Examples:**  
Node.js server, Django server, Spring Boot backend

---

### b. **Production Server**
- The **live environment** where actual users access the application.
- Must be secure, stable, and scalable.

---

### c. **Deployment (or Migration)**
- Moving code from a development/test environment to production.
- Includes backend code, frontend, configuration, and sometimes database migrations.

**Interview Questions:**
- What are some deployment strategies? (Blue-Green, Canary)
- How do you roll back a failed deployment?

---

## 9. **DBMS (Database Management System)**

**Definition:**  
A software system that manages access to and interaction with a database.

**Responsibilities:**
- Data creation, retrieval, update, deletion
- Security and concurrency control
- Backup and recovery

**Examples:**  
MySQL, MongoDB, PostgreSQL, Oracle DB

**Interview Questions:**
- What is the role of a DBMS?
- Explain the difference between DBMS and RDBMS.

---

## 10. **NoSQL (Non-Relational Databases)**

**Definition:**  
Flexible data models that don’t follow strict relational schemas.

### 📊 Types of NoSQL Databases

| Type           | Description                                    | Examples                      |
|----------------|------------------------------------------------|-------------------------------|
| **Key-Value**   | Simple key-value pairs                         | Redis, DynamoDB               |
| **Document**    | Stores documents (JSON, BSON)                  | MongoDB, CouchDB              |
| **Wide-Column** | Column-oriented storage                        | Cassandra, HBase              |
| **Graph**       | Nodes + edges to represent relationships       | Neo4j, ArangoDB               |
| **Search Engine** | Optimized for full-text search & analytics   | Elasticsearch, Solr           |

**Real-World Use Cases:**
- Key-Value: Session management (Redis)
- Document: Storing user profiles (MongoDB)
- Graph: Social networks (Neo4j)
- Search Engine DB: E-commerce product search (Elasticsearch)

**Interview Questions:**
- When would you choose MongoDB over MySQL?
- Explain CAP Theorem in context of NoSQL databases.

## 🔁 Bonus Concepts (Often Asked in Interviews)

### ACID Properties (for RDBMS):
- **Atomicity**: All or nothing
- **Consistency**: Always valid data
- **Isolation**: Operations don't interfere
- **Durability**: Data persists after success

### CAP Theorem (for NoSQL):
- **Consistency**
- **Availability**
- **Partition Tolerance**

Can't have all 3 at the same time in distributed systems.
