# Filter, Map, Reduce & Lambda advanced

- Reduce need to be called from functools
- e.g. from functools import reduce

```python
nums = [3,2,6,8,4,6,2,9]

evens = list(filter(lambda n:n%2==0, nums))
print(evens)
```

```python
nums = [3,2,6,8,4,6,2,9]

odd = list(filter(lambda n:n%2!=0, nums))
print(odd)
```

# Python Decorator
- It's an inheritance concept
- 

```python
# Production code
def div(a,b):
    print(a/b)

div(4,2)
```

```python
# New code # But we want the output exactly as 2.0
def div(a,b):
    print(a/b)

div(2,4)
```

```python
# the code should be
def div(a,b):
    if a<b:
        a,b=b,a
    print(a/b)

div(2,4)
```
