```python
import numpy as np
import pandas as pd

import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline

from sklearn import preprocessing
from sklearn.preprocessing import PolynomialFeatures
from sklearn.model_selection import train_test_split

from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import r2_score
```

```python
data = pd.read_csv(r"D:\NareshIT_PrakashSenapati\Prakash Sir-Notes\Auguest (Statistics)\Practicals\Car\car-mpg.csv")
data.head()
```

```python
data.info()
```

```python
# Drop the car name
# label the categorical value
#replace ? with nan
#replace all nan with median
data = data.drop(['car_name'], axis=1)
data['origin'] =data['origin'].replace({1: 'america', 2:'europe', 3:'asia'})
data=pd.get_dummies(data,columns = ['origin'], dtype=int)
data = data.replace('?',np.nan)
```

```python
data = data.apply(pd.to_numeric, errors='ignore')

#Fill missing values with median only for numeric columns
numeric_cols = data.select_dtypes(include=[np.number]).columns
data[numeric_cols] = data[numeric_cols].apply(lambda x : x.fillna(x.median()))
```

```python
data.head()
```

```python
X = data.drop(['mpg'], axis=1)#independent variable
y = data[['mpg']]
```

```python
#Feature Scaling, all the features should be on the similar scale


X_s = preprocessing.scale(X)
X_s = pd.DataFrame(X_s, columns=X.columns)

y_s = preprocessing.scale(y)
y_s = pd.DataFrame(y_s, columns=y.columns)
```
