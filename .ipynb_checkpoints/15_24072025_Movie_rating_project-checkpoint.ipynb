# 1. Dot notation (.)
## ✅ When it works:

- When the column name is a valid Python attribute name, i.e.,

- No spaces

- No special characters

- Doesn’t clash with DataFrame methods (e.g. count, sum).

## 🔴 When it fails:

- If column name has spaces, like Movie Title

- If column name conflicts with pandas methods, like count or sum

# 2. Bracket notation ([])
- ✅ When to use:

- Always works, regardless of:

- Spaces in column names

- Special characters

- Conflicts with method names

**✔️ Prefer bracket notation ([]) for production and general use to avoid hidden errors.**
